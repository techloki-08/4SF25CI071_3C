   2. Data Transformation Perform the required transformations, such as changing data types, dropping unnecessary columns, or adding new columns.


```python
# Safely drop 'Unnamed: 0' only if it exists in the DataFrame
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])
    print("Dropped 'Unnamed: 0' successfully.")
else:
    print("Column 'Unnamed: 0' is not present in this dataset. Skipping...")

print(df.head())
```

```python
# changing datatype
df['age']=df['age'].astype(float)
df['age'].head()
```

```python
# adding new column
df['total_engagement'] = df['library_visits_per_month'] + df['lms_logins_per_week']
df['risk_index'] = df['stress_level'] + (df['missed_deadlines'] * 2)
print("New columns added successfully!")
print(df[['person_id', 'total_engagement', 'risk_index']].head())
```

4.Handling Null Values Identify null values in the dataset. Handle null values using mean imputation wherever applicable.

```python
# checking null values
df.isnull().sum()
```

```python
# filling null values using mean for student dataset
df['age'] = df['age'].fillna(df['age'].mean())
df['admission_grade'] = df['admission_grade'].fillna(df['admission_grade'].mean())
df['gpa_semester_1'] = df['gpa_semester_1'].fillna(df['gpa_semester_1'].mean())
df['gpa_semester_2'] = df['gpa_semester_2'].fillna(df['gpa_semester_2'].mean())
df['attendance_rate'] = df['attendance_rate'].fillna(df['attendance_rate'].mean())
df['stress_level'] = df['stress_level'].fillna(df['stress_level'].mean())
```

```python
# checking null values after filling
df.isnull().sum()
```

```python
df['double_risk'] = df['risk_index'].apply(lambda x: x * 2)
print(df[['risk_index', 'double_risk']].head())
```

6.Data Visualization Visualize the generated insights using appropriate charts

```python
# counting students for each course type
course_counts = df.groupby('course_type')['person_id'].count()

# sorting in descending order
top_courses = course_counts.sort_values(ascending=False).head(10)
print(top_courses)
```

```python
# counting students for each nationality
nationality_counts = df.groupby('nationality')['person_id'].count()

# sorting in descending order
top_nationalities = nationality_counts.sort_values(ascending=False).head(10)
print(top_nationalities)
```

```python
# bar graph for top 10 nationalities by student count with 90-degree rotation
top_nationalities.plot(kind='bar')
plt.xlabel('Nationality', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.title('Top 10 Nationalities by Number of Students', fontsize=14, fontweight='bold')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# scatter plot for Semester 1 GPA vs Semester 2 GPA
plt.scatter(df['gpa_semester_1'], df['gpa_semester_2'])
plt.xlabel('GPA Semester 1')
plt.ylabel('GPA Semester 2')
plt.title('GPA Semester 1 vs GPA Semester 2')
plt.show()
```

```python
# scatter plot for LMS Logins per Week vs GPA Semester 1
plt.scatter(df['lms_logins_per_week'], df['gpa_semester_1'])
plt.xlabel('LMS Logins per Week')
plt.ylabel('GPA Semester 1')
plt.title('LMS Logins per Week vs GPA Semester 1')
plt.show()
```
